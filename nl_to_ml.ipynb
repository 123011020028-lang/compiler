# Import necessary modules

````python
import ipywidgets as widgets           
from IPython.display import display   
from IPython.display import Markdown   
````

# 🧠 Natural Language to Code Converter
### Enter a command in English and get Python code 

# Creating UI components

````python
input_text = widgets.Text(
    placeholder='Example: print numbers from 1 to 5',  
    description='Input:',                              
    layout=widgets.Layout(width='500px')               
)
convert_button = widgets.Button(
    description="Convert",          
    button_style='success'          
)
output_area = widgets.Output()
````

# Compiler function

````python
def nl_to_code(sentence):
    """
    This function performs a simple rule-based translation
    from natural language to Python code.
    
    Steps involved (Compiler Concept):
    1. Lexical Analysis  -> Tokenization
    2. Syntax Analysis   -> Pattern matching
    3. Semantic Analysis -> Understanding intent
    4. Code Generation   -> Producing Python code
    """
    
    # Convert sentence to lowercase for uniform processing
    sentence = sentence.lower()
    
    # Lexical Analysis: split sentence into tokens (words)
    tokens = sentence.split()

    # Loop (print numbers from X to Y)
    if "print" in tokens and "from" in tokens and "to" in tokens:
        try:
            # Extract start and end values
            start = int(tokens[tokens.index("from") + 1])
            end = int(tokens[tokens.index("to") + 1])

            # Generate Python loop code
            return f"""for i in range({start}, {end+1}):
    print(i)"""
        
        except:
            return "Error: Invalid numbers in input"

    # Addition (add X and Y)
    elif "add" in tokens:
        try:
            # Extract numbers from sentence
            nums = [int(word) for word in tokens if word.isdigit()]
            
            # Generate addition code
            return f"print({nums[0]} + {nums[1]})"
        
        except:
            return "Error: Could not find two numbers"

    # Simple print statement
    elif "print" in tokens:
        # Extract text after 'print'
        text = sentence.replace("print", "").strip()
        
        # Generate print statement
        return f'print("{text}")'

    # Default case: unknown command
    else:
        return "Command not recognized"
````

# Button Function

````python
def on_button_click(b):
    
    # Use output_area to display results
    with output_area:
        
        # Clear previous output
        output_area.clear_output()
        
        # Get user input
        user_input = input_text.value
        
        # Call compiler function
        result = nl_to_code(user_input)
        
        # Display formatted output as code block
        display(Markdown(f"### 💻 Generated Code:\n```python\n{result}\n```"))
````

# Functioning button

````python
convert_button.on_click(on_button_click)
````

# Display UI

````python
display(input_text)
display(convert_button)
display(output_area)
````
